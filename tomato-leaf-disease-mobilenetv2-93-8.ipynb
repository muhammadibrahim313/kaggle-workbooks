<p style="font-size:16px; line-height:1.7; text-align:justify;">
This dataset has been prepared for the classification of tomato leaf diseases using deep learning models. The dataset contains a comprehensive collection of tomato leaf images captured under different lighting and environmental conditions. Each image belongs to a specific class representing either a healthy leaf or one affected by a distinct disease. The primary objective is to develop a model capable of automatically identifying and classifying tomato plant diseases to assist farmers and researchers in early detection and prevention.
</p>

<h3 style="color:#c0392b;">Dataset Overview:</h3>
<ul style="font-size:15px; line-height:1.6;">
  <li><b>Total Number of Classes:</b> 10 (including healthy leaves)</li>
  <li><b>Image Resolution:</b> 256 × 256 pixels (resized for training consistency)</li>
  <li><b>Color Mode:</b> RGB (3-channel images)</li>
  <li><b>File Format:</b> JPEG/PNG</li>
  <li><b>Data Split:</b> Separate folders for training and validation</li>
  <li><b>Purpose:</b> Image classification and disease diagnosis in tomato plants</li>
</ul>

<h3 style="color:#c0392b;">Categories Included:</h3>
<ul style="font-size:15px; line-height:1.6;">
  <li><b>Tomato Bacterial Spot:</b> Small dark brown or black spots appearing on leaves, often with a yellow halo.</li>
  <li><b>Tomato Early Blight:</b> Irregular dark spots surrounded by concentric rings on lower leaves.</li>
  <li><b>Tomato Late Blight:</b> Large, irregular dark patches caused by fungal infection, often under humid conditions.</li>
  <li><b>Tomato Leaf Mold:</b> Yellowish spots on the upper surface of leaves with a grayish mold growth underneath.</li>
  <li><b>Tomato Septoria Leaf Spot:</b> Circular brown spots with dark borders appearing mainly on older leaves.</li>
  <li><b>Tomato Two-Spotted Spider Mite:</b> Fine webbing and small white or yellow spots caused by mite infestation.</li>
  <li><b>Tomato Target Spot:</b> Dark lesions with concentric rings that resemble a target pattern.</li>
  <li><b>Tomato Mosaic Virus:</b> Light and dark green mottling or mosaic pattern on the leaves.</li>
  <li><b>Tomato Yellow Leaf Curl Virus:</b> Leaf curling, yellowing, and stunted growth caused by viral infection.</li>
  <li><b>Healthy Tomato Leaves:</b> Uniform green leaves showing no signs of disease or discoloration.</li>
</ul>

<p style="font-size:16px; line-height:1.7; text-align:justify;">
This dataset serves as an essential resource for developing robust computer vision models in agricultural disease detection. By accurately classifying tomato leaf diseases, it can support farmers in timely intervention, improve crop yield, and reduce economic losses caused by late disease diagnosis.
</p>

```python
%time
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
import os



# Paths
train_path = "/kaggle/input/tomatoleaf/tomato/train"
val_path = "/kaggle/input/tomatoleaf/tomato/val"

# Train dataset
train_ds = tf.keras.utils.image_dataset_from_directory(
    train_path,
    labels="inferred",
    label_mode="int",
    image_size=(256, 256),
    batch_size=32
)

# Validation dataset
val_ds = tf.keras.utils.image_dataset_from_directory(
    val_path,
    labels="inferred",
    label_mode="int",
    image_size=(256, 256),
    batch_size=32
)
```

Dataset successfully loaded.

- Training set: 10,000 images belonging to 10 classes.
- Validation set: 1,000 images belonging to the same 10 classes.
- Each image is resized to 256×256 pixels.
- Images are batched with a batch size of 32.

```python
%time
# sample image
for images, labels in train_ds.take(1):
    sample_img = images[0].numpy().astype("uint8")
    break

plt.title("Sample Train Image (Before Scaling)")
plt.imshow(sample_img)
plt.axis("off")
plt.show()

print("Pixel values (Red channel, Before Scaling):")
for row in sample_img[:, :, 0]:
    for pixel in row[:20]:   
        print("{:3}".format(int(pixel)), end=" ")
    print()
# Scale datasets (0–1 range)
train_ds = train_ds.map(lambda x, y: (x/255.0, y))
val_ds = val_ds.map(lambda x, y: (x/255.0, y))


sample_img_scaled = sample_img / 255.0

plt.title("Sample Train Image (After Scaling)")
plt.imshow(sample_img_scaled)
plt.axis("off")
plt.show()

print("\nPixel values (Red channel, After Scaling):")
for row in sample_img_scaled[:, :, 0]:
    for pixel in row[:20]:
        print("{:.2f}".format(pixel), end=" ")
    print()
```

Before scaling, the image pixels were in the 0–255 range (uint8 RGB values). After normalization, the same pixels were converted into the 0–1 floating-point range. Importantly, the image appearance did not change, only the pixel value range was adjusted  which ensures standardized inputs, stable gradients, and better training efficiency.

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.shuffle(1000).cache().prefetch(buffer_size=AUTOTUNE)
val_ds   = val_ds.cache().prefetch(buffer_size=AUTOTUNE)
```

* The training and validation datasets have been optimized using cache(), shuffle(), and prefetch().

* shuffle(1000) ensures better randomness during training.

* cache() stores data in memory after the first load, reducing disk I/O.

* prefetch(AUTOTUNE) overlaps data preprocessing with model execution for faster training.

Overall, this step improves training efficiency, reduces latency, and ensures a smooth data pipeline.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt

# Base Model MobileNetV2
base_model = keras.applications.MobileNetV2(
    input_shape=(256, 256, 3),
    include_top=False,
    weights="imagenet"
)

# freeze 
base_model.trainable = False

# Head (classifier)
x = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu", kernel_regularizer=regularizers.l2(1e-4)),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(len(class_names), activation="softmax")
])

model = x
```

```python
# Compile
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Training 
early_stop = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)
lr_reducer = ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, verbose=1)

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=50,
    callbacks=[early_stop, lr_reducer],
    verbose=1
)

# Fine-tuning stage
base_model.trainable = True
# Lekin sirf last kuch layers ko unfreeze karte hain
for layer in base_model.layers[:-40]:  # last 40 layers trainable
    layer.trainable = False

# Re-compile with smaller learning rate
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=50,
    callbacks=[early_stop, lr_reducer],
    verbose=1
)
```

The model was trained in two phases first, the pretrained MobileNetV2 layers were frozen to train only the classifier (feature extraction phase). Then, in the second phase, the last few layers

```python
# Combine history
acc = history1.history['accuracy'] + history2.history['accuracy']
val_acc = history1.history['val_accuracy'] + history2.history['val_accuracy']
loss = history1.history['loss'] + history2.history['loss']
val_loss = history1.history['val_loss'] + history2.history['val_loss']

# final scores
print("="*60)
print("Model Training Summary")
print("="*60)
print(f"Total Epochs Trained: {len(acc)}")
print(f"Final Training Accuracy:   {acc[-1]*100:.2f}%")
print(f"Final Validation Accuracy: {val_acc[-1]*100:.2f}%")
print(f"Final Training Loss:       {loss[-1]:.4f}")
print(f"Final Validation Loss:     {val_loss[-1]:.4f}")
print("="*60)
```
