## **Main Goal**

The main goal of this project is to analyze the Unreal Engine dataset to gain insights into its evolution over time, its usage patterns, and the trends in game development associated with it.

## **Background**

Unreal Engine, developed by Epic Games, is one of the most popular game engines in the world, known for its powerful features and versatility. This dataset provides a comprehensive collection of information about Unreal Engine, ranging from its historical versions to the latest updates and features.

## **Dataset Information**

The dataset contains various attributes and metrics related to Unreal Engine, including but not limited to:

- Historical versions and release dates
- Features introduced in each version
- Usage statistics and trends
- Community feedback and discussions

This dataset is sourced from Kaggle and is an invaluable resource for understanding the evolution of game development technology, particularly within the Unreal Engine ecosystem.

**Dataset Link:** [Unreal Engine - From Past to Future](https://www.kaggle.com/datasets/middlehigh/unreal-engine-from-past-to-future)

## **Analysis Plan**

Our analysis will focus on:

- **Version Evolution:** Analyzing how Unreal Engine has evolved over time in terms of features, performance, and community adoption.
- **Usage Patterns:** Identifying trends in how developers utilize different versions of the engine for various types of projects.
- **Community Insights:** Exploring community discussions and feedback to understand user sentiments and emerging needs.
